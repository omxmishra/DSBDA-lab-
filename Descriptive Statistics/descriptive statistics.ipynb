```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df.groupby('species')['sepal_length'].agg(['mean','median','min','max','std'])
```

```python
df.groupby('species')['sepal_length'].apply(list)
```

```python
setosa = df[df['species'] == 'setosa']
versicolor = df[df['species'] == 'versicolor']
virginica = df[df['species'] == 'virginica']
```

```python
print("Setosa:\n", setosa.describe())
print("\nVersicolor:\n", versicolor.describe())
print("\nVirginica:\n", virginica.describe())
```

```python
print("Setosa Percentiles:\n", setosa.select_dtypes(include='number').quantile([0.25, 0.5, 0.75]))
print("\nVersicolor Percentiles:\n", versicolor.select_dtypes(include='number').quantile([0.25, 0.5, 0.75]))
print("\nVirginica Percentiles:\n", virginica.select_dtypes(include='number').quantile([0.25, 0.5, 0.75]))
```
