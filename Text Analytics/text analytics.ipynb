```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
```

```python
text = "Natural Language Processing is a field of Artificial Intelligence that deals with text data."
print("Original Text:", text)
```

```python
tokens = word_tokenize(text)
print("\nTokens:", tokens)
```

```python
pos_tags = pos_tag(tokens)
print("\nPOS Tags:", pos_tags)
```

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w.lower() not in stop_words]
print("\nAfter Stopword Removal:", filtered_tokens)
```

```python
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered_tokens]
print("\nStemmed:", stemmed)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]
print("\nLemmatized:", lemmatized)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

clean_tokens = [w.lower().strip('.,') for w in filtered_tokens]
doc = " ".join(clean_tokens)
```

```python
vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform([doc])
print("\nWords:", vectorizer.get_feature_names_out())
print("TF-IDF:\n", tfidf.toarray())
```
